```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.stattools import adfuller
```

```python
result = adfuller(df_dow_jones['Open'])

print('ADF Statistic: %f' % result[0])

print('p-value: %f' % result[1])
```

```python
from statsmodels.graphics.tsaplots import plot_acf

# Gráfico de autocorrelación para la columna 'Open' de df_dow_jones
plot_acf(df_dow_jones['Open'], lags=50)
plt.title('Autocorrelación - Dow Jones (Open)')
plt.show()
```

```python
from statsmodels.graphics.tsaplots import plot_pacf

# Gráfico de autocorrelación parcial para la columna 'Open' de df_dow_jones
plot_pacf(df_dow_jones['Open'], lags=10, method='ywm')
plt.title('Autocorrelación Parcial - Dow Jones (Open)')
plt.show()
```

```python

import lightgbm as lgb
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Asegúrate de que la columna 'Date' esté en formato datetime
df_dow_jones['Date'] = pd.to_datetime(df_dow_jones['Date'])

# Ordenar por fecha
df_dow_jones = df_dow_jones.sort_values(by='Date')



# Crear características temporales
df_dow_jones['dayofweek'] = df_dow_jones['Date'].dt.dayofweek  # 0=Lunes
df_dow_jones['month'] = df_dow_jones['Date'].dt.month           # 1–12

# Codificación cíclica
df_dow_jones['dayofweek_sin'] = np.sin(2 * np.pi * df_dow_jones['dayofweek'] / 7)
df_dow_jones['dayofweek_cos'] = np.cos(2 * np.pi * df_dow_jones['dayofweek'] / 7)

df_dow_jones['month_sin'] = np.sin(2 * np.pi * df_dow_jones['month'] / 12)
df_dow_jones['month_cos'] = np.cos(2 * np.pi * df_dow_jones['month'] / 12)

# Crear características de lag (por ejemplo, 3 lags)
lags = 3
for lag in range(1, lags + 1):
    df_dow_jones[f'Open_lag_{lag}'] = df_dow_jones['Open'].shift(lag)

# Eliminar filas con valores nulos
df_dow_jones = df_dow_jones.dropna()

# Definir las características (X) y la variable objetivo (y)
X = df_dow_jones[
    [f'Open_lag_{lag}' for lag in range(1, lags + 1)] + 
    ['dayofweek_sin', 'dayofweek_cos', 'month_sin', 'month_cos']
]
y = df_dow_jones['Open']

# 95% para entrenamiento, 5% para prueba
train_size = int(len(X) * 0.95)

X_train = X.iloc[:train_size]
X_test = X.iloc[train_size:]

y_train = y.iloc[:train_size]
y_test = y.iloc[train_size:]

# Crear el modelo
model = lgb.LGBMRegressor(random_state=30)
model.fit(X_train, y_train)

# Predicciones
y_pred = model.predict(X_test)

# Graficar
plt.figure(figsize=(12, 6))
plt.plot(df_dow_jones['Date'].iloc[train_size:], y_test, label='Datos reales', color='blue')
plt.plot(df_dow_jones['Date'].iloc[train_size:], y_pred, label='Predicciones', linestyle='--', color='red')
plt.xlabel('Fecha')
plt.ylabel('Precio de Apertura (Open)')
plt.title('Predicciones con variables temporales cíclicas')
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Evaluación
print("R^2 score en el conjunto de prueba:", model.score(X_test, y_test))



```

```python

import lightgbm as lgb
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

lasps=df_dow_jones.copy()

# Asegúrate de que la columna 'Date' esté en formato datetime
lasps['Date'] = pd.to_datetime(lasps['Date'])

# Ordenar por fecha
lasps= lasps.sort_values(by='Date')



# Crear características temporales
lasps['dayofweek'] = lasps['Date'].dt.dayofweek  # 0=Lunes
lasps['month'] = lasps['Date'].dt.month           # 1–12

# Codificación cíclica
lasps['dayofweek_sin'] = np.sin(2 * np.pi * lasps['dayofweek'] / 7)
lasps['dayofweek_cos'] = np.cos(2 * np.pi * lasps['dayofweek'] / 7)

lasps['month_sin'] = np.sin(2 * np.pi * lasps['month'] / 12)
lasps['month_cos'] = np.cos(2 * np.pi * lasps['month'] / 12)

# Crear características de lag (por ejemplo, 3 lags)
lags = 5
for lag in range(1, lags + 1):
    lasps[f'Open_lag_{lag}'] = lasps['Open'].shift(lag)

# Eliminar filas con valores nulos
lasps = lasps.dropna()

# Definir las características (X) y la variable objetivo (y)
X = lasps[
    [f'Open_lag_{lag}' for lag in range(1, lags + 1)] + 
    ['dayofweek_sin', 'dayofweek_cos', 'month_sin', 'month_cos']
]
y = lasps['Open']

# 95% para entrenamiento, 5% para prueba
train_size = int(len(X) * 0.95)

X_train = X.iloc[:train_size]
X_test = X.iloc[train_size:]

y_train = y.iloc[:train_size]
y_test = y.iloc[train_size:]

# Crear el modelo
model = lgb.LGBMRegressor(random_state=20)
model.fit(X_train, y_train)

# Predicciones
y_pred = model.predict(X_test)

# Graficar
plt.figure(figsize=(12, 6))
plt.plot(lasps['Date'].iloc[train_size:], y_test, label='Datos reales', color='blue')
plt.plot(lasps['Date'].iloc[train_size:], y_pred, label='Predicciones', linestyle='--', color='red')
plt.xlabel('Fecha')
plt.ylabel('Precio de Apertura (Open)')
plt.title('Predicciones con variables temporales cíclicas')
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Evaluación
print("R^2 score en el conjunto de prueba:", model.score(X_test, y_test))

# Obtener la importancia de las características
importance_df = pd.DataFrame({
    'Feature': X.columns,
    'Importance': model.feature_importances_
}).sort_values(by='Importance', ascending=False)

# Mostrar tabla
print("\nImportancia de características:")
print(importance_df)

# Graficar
plt.figure(figsize=(10, 6))
plt.barh(importance_df['Feature'], importance_df['Importance'], color='teal')
plt.xlabel('Importancia')
plt.ylabel('Característica')
plt.title('Importancia de características (LightGBM)')
plt.gca().invert_yaxis()  # La más importante arriba
plt.tight_layout()
plt.show()
```
