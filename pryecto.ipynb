```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.stattools import adfuller
```

```python
result = adfuller(df_dow_jones['Open'])

print('ADF Statistic: %f' % result[0])

print('p-value: %f' % result[1])
```

```python
from statsmodels.graphics.tsaplots import plot_acf

# Gráfico de autocorrelación para la columna 'Open' de df_dow_jones
plot_acf(df_dow_jones['Open'], lags=50)
plt.title('Autocorrelación - Dow Jones (Open)')
plt.show()
```

```python
from statsmodels.graphics.tsaplots import plot_pacf

# Gráfico de autocorrelación parcial para la columna 'Open' de df_dow_jones
plot_pacf(df_dow_jones['Open'], lags=10, method='ywm')
plt.title('Autocorrelación Parcial - Dow Jones (Open)')
plt.show()
```

```python
from sklearn.model_selection import train_test_split

# Dividir los datos en entrenamiento (80%) y prueba (20%)
train_dow_jones, test_dow_jones = train_test_split(df_dow_jones, test_size=0.2, random_state=42)

print("Entrenamiento:")
print(train_dow_jones.head())
print("\nPrueba:")
print(test_dow_jones.head())
```
